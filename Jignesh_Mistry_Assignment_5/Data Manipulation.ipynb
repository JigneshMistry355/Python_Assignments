### 1. handle missing values

```python
df = pd.read_csv('yearly_sales_2024_2.csv')
df.head(5)
```

##### Drop rows with missing values

```python
df.dropna()
```

```python
# Get rows with missing values 

missing_values_data = df[df.isna().any(axis=1)]
missing_values_data
```

```python
# Get the price of webcam 


webcam_price = df.loc[(df["Product"] == 'Webcam') & (df["Price"].notna()), "Price"].iloc[0]
webcam_price
```

```python
# Edit missing values for webcam

df.loc[(df["Product"] == 'Webcam') & (df["Price"].isna()), "Price"] = webcam_price
```

```python
# Check if Price column of webcam is still empty


df[df.isna().any(axis=1)]
```

```python
# Fill missing value in Quantity column for webcam


df.loc[(df["Product"] == 'Webcam') & (df["Quantity"].isna()), "Quantity"] = \
df.loc[(df["Product"] == 'Webcam') & (df["Quantity"].isna())].apply(lambda row: row["Price"]/webcam_price, axis=1)
```

```python
# Check if missing Quantity is present for webcam


df[df.isna().any(axis=1)]
```

```python
for idx, row in missing_values_data.iterrows():
    product = row["Product"]
    unit_price_series = df.loc[(df["Product"] == product) & (df["Price"].notna()), "Price"]
    if not unit_price_series.empty:
        unit_price = unit_price_series.iloc[0]
        if pd.isna(row["Quantity"]) & pd.notna(row["Price"]):
            df.at[idx, "Quantity"] = row["Price"] / unit_price
```
