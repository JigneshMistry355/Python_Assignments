```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import streamlit as st
```

```python
df = pd.read_csv("heart.csv")
df.head(10)
```

---

```python
sns.histplot(data=df, x="age")
plt.title("Count of petients by age")
plt.show()
```

---

```python
sns.countplot(data=df, x=df['sex']==1)
plt.title("Female Vs Male count")
plt.show()
```

```python
sns.boxplot(x=df["age"] > 40, y="chol", data=df)
plt.title("Cholestrol in people more than age 40")
plt.show()
```

### Cholestrol in Female vs Male

---

```python
sns.boxplot(x=df['sex'] == 1, y='chol', data=df)
plt.title("Female Vs Male")
plt.show()
```

### Age Vs Cholestrol

```python
sns.scatterplot(data=df, x="age", y="chol")
plt.show()
```

```python
sns.barplot(data=df, x="sex", y='target', estimator="mean")
plt.show()
```

```python
sns.barplot(data=df, x="cp", y='chol', estimator='mean')
plt.show()
```
